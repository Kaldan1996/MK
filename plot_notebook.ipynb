# Amino Acid Hydrophobicity vs IDP Propensity Plot

This notebook creates a scatter plot where amino acid one-letter codes are used as markers.

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
# Load your data
# Replace 'example_data.csv' with your actual data file
df = pd.read_csv('example_data.csv')
df.head()
```

```python
# If you want to load from your own data, uncomment and modify:
# df = pd.read_csv('your_data.csv')
# Specify your column names:
hydrophobicity_col = 'hydrophobicity'
idp_col = 'IDP_propensity'
aa_col = 'amino_acid'
```

```python
# Extract data
x = df[hydrophobicity_col].values
y = df[idp_col].values
labels = df[aa_col].values

# Create figure and axis
fig, ax = plt.subplots(figsize=(12, 8))

# Plot each amino acid as a text marker
for i in range(len(x)):
    ax.text(x[i], y[i], labels[i],
            ha='center', va='center',
            fontsize=12, fontweight='bold',
            bbox=dict(boxstyle='circle,pad=0.3',
                     facecolor='lightblue',
                     edgecolor='navy',
                     alpha=0.7))

# Set labels and title
ax.set_xlabel('Hydrophobicity', fontsize=14, fontweight='bold')
ax.set_ylabel('IDP Propensity', fontsize=14, fontweight='bold')
ax.set_title('Amino Acid Hydrophobicity vs IDP Propensity', fontsize=16, fontweight='bold', pad=20)

# Add grid for better readability
ax.grid(True, alpha=0.3, linestyle='--')

# Adjust layout
plt.tight_layout()

# Save the plot
plt.savefig('amino_acid_plot.png', dpi=300, bbox_inches='tight')
print("Plot saved as amino_acid_plot.png")

# Display the plot
plt.show()
```

## Customization Options

You can customize the appearance by modifying:
- `figsize`: Change plot dimensions
- `fontsize`: Adjust text size for amino acid letters
- `facecolor`: Change the background color of the markers
- `edgecolor`: Change the border color of the markers
- `boxstyle`: Try 'round,pad=0.3', 'square,pad=0.3', etc.

```python
# Alternative version with different styling
fig, ax = plt.subplots(figsize=(14, 10))

# Define color map based on hydrophobicity or IDP propensity
from matplotlib.cm import RdYlBu
from matplotlib.colors import Normalize

# Normalize colors based on hydrophobicity
norm = Normalize(vmin=x.min(), vmax=x.max())
colors = RdYlBu(norm(x))

# Plot each amino acid with color coding
for i in range(len(x)):
    ax.text(x[i], y[i], labels[i],
            ha='center', va='center',
            fontsize=14, fontweight='bold',
            bbox=dict(boxstyle='circle,pad=0.35',
                     facecolor=colors[i],
                     edgecolor='black',
                     linewidth=1.5,
                     alpha=0.8))

# Set labels and title
ax.set_xlabel('Hydrophobicity', fontsize=14, fontweight='bold')
ax.set_ylabel('IDP Propensity', fontsize=14, fontweight='bold')
ax.set_title('Amino Acid Hydrophobicity vs IDP Propensity (Color-coded)', 
             fontsize=16, fontweight='bold', pad=20)

# Add grid
ax.grid(True, alpha=0.3, linestyle='--')

# Add colorbar
sm = plt.cm.ScalarMappable(cmap=RdYlBu, norm=norm)
sm.set_array([])
cbar = plt.colorbar(sm, ax=ax)
cbar.set_label('Hydrophobicity', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.savefig('amino_acid_plot_colored.png', dpi=300, bbox_inches='tight')
print("Colored plot saved as amino_acid_plot_colored.png")
plt.show()
```
